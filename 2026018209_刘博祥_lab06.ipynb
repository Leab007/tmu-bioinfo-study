### 2.3　体表面积
化疗药物常按体表面积（BSA，单位 m²）给药。Mosteller 公式：

$$BSA = \sqrt{\frac{身高(cm) \times 体重(kg)}{3600}}$$

写函数 `bsa(height_cm, weight_kg)`，返回体表面积（不要四舍五入）。开平方可以用 `math.sqrt(…)`，也可以用 `** 0.5`。

```python
def bsa(height_cm, weight_kg):
    # ↓↓↓ 在这里写代码（删掉 pass）↓↓↓
    return (height_cm*weight_kg/3600)**0.5
    # ↑↑↑

check("2.3 bsa", [bsa(170, 65), bsa(160, 52)])
```

### 3.1　平均值与标准差
1. `mean(data)`：用循环求平均值（**不要用 `sum`**）
2. `sd(data)`：样本标准差，公式 $\sqrt{\frac{\sum (x - \bar{x})^2}{n - 1}}$。**在 sd 里调用 mean**，不要重新写一遍求平均的循环。

写完后和 `statistics.stdev(glucose)` 对照。

```python
glucose = [5.6, 6.1, 5.9, 7.2, 6.0, 5.8, 6.3,
           7.5, 6.4, 5.7, 6.8, 7.1, 6.2, 5.9]
def mean(data):
    # ↓↓↓ 在这里写代码（删掉 pass）↓↓↓
    glu,n=0,0
    for g in glucose:
        glu+=g
        n+=1
    return glu/n
    # ↑↑↑
temps = [36.8, 38.5, 37.1, 39.2, 36.5, 37.6, 38.0, 36.9]

def sd(data):
    # ↓↓↓ 在这里写代码（删掉 pass）↓↓↓
    m=mean(data)
    s=0
    for t in data:
        s+=(t-m)**2
    return (s/(len(data)-1))**0.5
        
    # ↑↑↑

check("3.1 mean", mean(glucose))
check("3.1 sd", sd(glucose))
```

### 3.3　药物代谢：几小时降到安全线？
把第 5 讲的 while 循环写成函数 `hours_to_safe(c0, half_life, safe=10)`：
初始浓度 `c0`，每过 `half_life` 小时浓度减半，返回浓度**第一次低于** `safe` 时经过的小时数。

check 测三种情况：(100, 4)；(120, 6, 安全线 8)；以及一开始就已经低于安全线的 (5, 4)——应该返回 0。

```python
def hours_to_safe(c0, half_life, safe=10):
    # ↓↓↓ 在这里写代码（删掉 pass）↓↓↓
    h=0
    while c0>=safe:
        c0/=2
        h+=half_life
    return h
    # ↑↑↑

check("3.3 hours_to_safe", [hours_to_safe(100, 4), hours_to_safe(120, 6, 8), hours_to_safe(5, 4)])
```

### 4.1　三块积木
照着课件第 18 页，写出：
- `sigmoid(z)`：返回 $\frac{1}{1 + e^{-z}}$（用 `math.exp`）
- `weighted_sum(x, w, b)`：x、w 是等长列表，返回 b + Σ x[i]·w[i]（用循环）
- `neuron(x, w, b)`：**调用前两个函数**，返回神经元的输出

```python
def sigmoid(z):
    # ↓↓↓ 在这里写代码（删掉 pass）↓↓↓
    return 1/(1+math.exp(-z))
    # ↑↑↑

def weighted_sum(x, w, b):
    # ↓↓↓ 在这里写代码（删掉 pass）↓↓↓
    z=b
    for i in range(len(x)):
        z=z+w[i]*x[i]
    return z
    # ↑↑↑

def neuron(x, w, b):
    # ↓↓↓ 在这里写代码（删掉 pass）↓↓↓
    return sigmoid(weighted_sum(x,w,b))
    # ↑↑↑


check("4.1 神经元积木", [sigmoid(0),
                        weighted_sum([1.7, 1.0], [0.9, 0.6], -1.5),
                        neuron([1.7, 1.0], [0.9, 0.6], -1.5)])
```
