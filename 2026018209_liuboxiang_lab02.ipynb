```python
'''
文件名称：EXEC07_buquan01.py
题目内容：输入一个正整数，判断其奇偶性。
'''
#输入部分
x=(eval)(input("请输入一个正整数："))
#处理部分和输出部分
if(x%2==0):
    print(x,"是偶数")
if (x%2!=0):
    print(x,"是奇数")
```

```python
'''
文件名称：EXEC07_buquan02.py
题目内容：输入三角形的边长，判断这三条边能否组成一个三角形，并输出判定结果。
'''
#输入部分
a=eval(input("请输入第一条线段长度："))
b=eval(input("请输入第二条线段长度："))
c=eval(input("请输入第三条线段长度："))
#处理部分和输出部分
if a+b>c and a-b<c:
    print("Yes")
else:
    print("No")
```

```python
'''
文件名称：EXEC07_buquan03.py
题目内容：输入一个年份，判断这一年是否是闰年。（四年一闰，百年不闰，四百年再闰）。
'''
#输入部分
year=eval(input("请输入一个年份："))
#处理部分和输出部分
if year%4==0 and not year%100==0 or year%400==0:
    print("是闰年")
else:
    print("不是闰年")
```

```python
'''
文件名称:"EXEC07_gaicuo01.py"
题目内容：从键盘输入一个数，判断其是否是 5 的倍数而不是 7 的倍数。如果是，输出 "Yes"，否则输出 "No"。
'''
#输入部分
num=eval(input("请输入一个数："))
#处理部分和输出部分
if num%5==0 and num%7!=0:
    print("Yes")
else:
    print("No")
```

```python
'''
文件名称:"EXEC07_gaicuo02.py"
1.题目内容：根据北京市轨道交通的计程限时票制，从键盘输入公里数，请输出对应的票价。
2.题目说明：北京市轨道交通（除首都机场线、大兴机场线外）实行计程限时票制，具体票价方案：
                     起步 6 公里（含）内 3 元
                     6 公里至 12 公里（含） 4  元
                     12 公里至 22 公里（含） 5 元
                     22 公里至 32 公里（含） 6 元
                     32 公里以上部分，每增加 1 元可乘坐 20 公里。
'''
#输入部分
s=eval(input("请输入一个公里数："))
#处理部分
import math
if s>32:
    price=6+(math.ceil((s-32)/20))*1
elif s<=12: 
    price=4
elif s<=22:
     price=5
elif s<=32:
    price=6
else:
    price=3
#输出部分
print("票价为"+str(price)+"元")
```

```python
#练习1
weight = float(input("体重（kg）："))
height = float(input("身高(m) :"))
who= weight/height**2
if who<18.5:
    print("偏瘦")
elif who<25:
    print("正常")
elif who<30:
    print("超重")
else:
    print("肥胖")
```

```python
#练习2
pressure= float(input("收缩压(mmhg):"))
if pressure<120:
    print("正常")
elif pressure<140:
    print("正常高值")
elif pressure>=140:
    print("高血压")
```
