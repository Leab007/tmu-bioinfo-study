### 3.2　一段 DNA：四种碱基各几个，GC 含量多少
用**一个** for 循环，同时完成：
- 四个计数器 `nA, nT, nG, nC`，最后写成 `counts = [nA, nT, nG, nC]`
- GC 含量 `gc`：G 和 C 占全部碱基的比例

提示：循环里用 `if … elif … elif … else`（第 4 讲）。

````python
seq = "ATGGCGTACGTTAGCCGATAAGCTTGCACG"
counts = None
gc = None
# ↓↓↓ 在这里写代码 ↓↓↓
nA=0
nG=0
nC=0
nT=0
for base in seq:
    if base=='A':
        nA= nA+1
    elif base=='G':
        nG+=1
    elif base=='C':
        nC+=1
    else :
        nT+=1
counts = [nA,nT,nG,nC]
gc=(nG+nC)/len(seq)
# ↑↑↑
check("3.2 四种碱基个数 [A, T, G, C]", counts)
check("3.2 GC 含量", gc)
````

### 3.3　互补链：用循环代替 replace
第三章说过：连续用 `replace` 求互补链，会“换过去又换回来”。用 for + if 逐个碱基处理就没有这个问题：**A↔T，G↔C**。
- `comp`：互补链（例：`"ATGC"` → `"TACG"`）
- `rev_comp`：反向互补链（把 `comp` 倒过来，用切片）

提示：先写 `comp = ""`，循环里每次 `comp = comp + "T"` 这样往后接。

````python
seq = "ATGGCGTACGTTAGCC"
comp = None
rev_comp = None
# ↓↓↓ 在这里写代码 ↓↓↓
if comp is None:
    comp=""
for X in seq:
    if X == 'A':
        new_X='T'
    elif X=='G':
        new_X='C'
    elif X=='C':
        new_X='G'
    else:
        new_X='A'
    comp=comp+new_X
rev_comp=comp[::-1]


# ↑↑↑
check("3.3 互补链", comp)
check("3.3 反向互补链", rev_comp)
````

### 3.4　神经元：三项体征的加权求和
一位病人：体温 39.0℃，心率 110 次/分，呼吸 26 次/分。每项先换算成“超出判断线多少”：

| 体征 | 实测 | 判断线 | 换算单位 | 输入 x |
|---|---|---|---|---|
| 体温 | 39.0 | 37.3 | 1 ℃ | (39.0 − 37.3) / 1 = 1.7 |
| 心率 | 110 | 100 | 10 次 | (110 − 100) / 10 = 1.0 |
| 呼吸 | 26 | 20 | 5 次 | ？ |

权重 `w = [0.9, 0.6, 0.8]`，偏置 `b = -1.5`。

要求：用**一个** for 循环，每一轮先算出第 i 项的输入 `x`，再把 `w[i] * x` 累加进 `z`（`z` 从 `b` 开始）；循环结束后算出 `y = 1 / (1 + math.exp(-z))`。

````python
raw   = [39.0, 110, 26]    # 实测值
line  = [37.3, 100, 20]    # 判断线
scale = [1, 10, 5]         # 换算单位
w = [0.9, 0.6, 0.8]
b = -1.5
z = None
y = None
# ↓↓↓ 在这里写代码 ↓↓↓
import math
if z is None:
    z=b
for i in range(len(raw)):
    X=(raw[i]-line[i])/scale[i]
    z=z+w[i]*X
y=1/(1+math.exp(-z))

# ↑↑↑
check("3.4 加权求和 z", z)
check("3.4 预测 y", y)
````
